```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd
```

```python
df = pd.read_csv("sales_record.csv")
df
```

```python
df.info()
```

```python
## comparition - bar chart
```

```python
dfg = df.groupby("Sales Channel").agg({"Total Profit": "sum"}).reset_index()
dfg
```

```python
plt.bar(x = "Sales Channel",  height = "Total Profit", data = dfg)
```

```python
sns.barplot(x = "Sales Channel", y = "Total Profit", data = dfg)
```

```python
## comparition - multiline chart
```

```python
df.info()
```

```python
plt.figure(figsize = (12,6))
plt.xticks(rotation = 90)
sns.lineplot(x = "Order Date", y = "Total Profit", data = df, label = "Profit")
sns.lineplot(x = "Order Date", y = "Total Revenue", data = df, label = "Total Revenue")
plt.legend()
```

```python
sns.histplot(x = "tenure", data = df)
```

```python
df_final = dfg.sort_values(by = "MonthlyCharges", ascending = False)
df_final
```

```python
sns.barplot(x = "PaymentMethod", y = "MonthlyCharges", data = df_final)
```
