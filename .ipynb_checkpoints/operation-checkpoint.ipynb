```python
def max_of_2(a,b):
    if a > b:
        result  = a
    else:
        result = b
    return result
```

```python
max_of_2(1,5)
```

```python
max_of_2(-5,-7)
```

```python
def odd_even(a):
    if a%2 == 0:
        print("a is even")
    else:
        print("a is odd")
```

```python
odd_even(23129)
```

```python
def power(a,b):
    result = a
    for i in range(b - 1):
        result = result * a
    return result
```

```python
power(3,2)
```

```python
def sum_of_list(a):
    result = 0
    for i in a:
        result = result + i
    return result
```

```python
sum_of_list([100,30,20,400])
```
