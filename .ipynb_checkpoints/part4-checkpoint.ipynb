```python
#t Test
```

```python
import scipy.stats as stats
```

```python
import numpy as np
```

```python
data = np.random.randint(20,25,30)
data
```

```python
## on sided - Right (Greater)
## h0 = u = 20
## h1 = u > 20
```

```python
ttest = stats.ttest_1samp(a=data, popmean = 20, alternative = "greater")
ttest
```

```python
pvalue = ttest.pvalue
print(pvalue)
```

```python
alpha = 0.05
print(pvalue < alpha)
```

```python
print("conclution: We success to reject H0. u > 20")
```

```python
## on sided - Left (less)
## h0 = u = 20
## h1 = u < 20
```

```python
ttest = stats.ttest_1samp(a=data, popmean = 20, alternative = "less")
ttest
```

```python
pvalue = ttest.pvalue
alpha = 0.05
print(pvalue < alpha)
```

```python
print("conclution: We failed to reject H0. u = 20")
```

```python
## two sided
## h0 = u = 20
## h1 = u != 20
```

```python
ttest = stats.ttest_1samp(a=data, popmean = 20)
ttest
```

```python
pvalue = ttest.pvalue
alpha = 0.05
print(pvalue < alpha)
```

```python
print("conclution: We success to reject H0. u != 20")
```
