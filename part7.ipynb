Additional Data Preparation Before modeling

```python
import pandas as pd
```

```python
df = pd.read_csv("flight_price_data.csv")
df
```

```python
## one hot encoding
## adalah proses mengubah variabel kategorikal menjadi variabel numerik

df.info()
```

```python
df_obj = df.select_dtypes(["object"])
col_obj = df_obj.columns
col_obj
```

```python
df_clean = pd.get_dummies(df, columns=col_obj)
df_clean
```

```python
## train test split
## adalah proses membagi dataset menjadi dua bagian, yaitu data latih dan data uji
## data train digunakan untuk melatih model, sedangkan data test digunakan untuk menguji model
from sklearn.model_selection import train_test_split
## y adalah target dan x adalah input
y = df_clean["price"]
x = df_clean.drop(columns=["price"])
```

```python
y
```

```python
x
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(x, y, test_size= 0.3, random_state=1)
```

```python
print(X_train.shape)
print(X_test.shape)
print(Y_train.shape)
print(Y_test.shape)
```

```python
## regression case study
from sklearn import linear_model
```

```python
linear_reg = linear_model.LinearRegression()

linear_reg.fit(X_train, Y_train)
```

```python
print("intercept:", linear_reg.intercept_)
```

```python
##evaluasi model
from sklearn.metrics import root_mean_squared_error, mean_absolute_error, r2_score

y_pred = linear_reg.predict(X_test)
```

```python
mean_absolute_error(Y_test, y_pred)
```

```python
root_mean_squared_error(Y_test, y_pred)
```

```python
import numpy as np
print(np.mean(Y_test))
print(np.std(Y_test))
```

```python
r2_score(Y_test, y_pred)
```
