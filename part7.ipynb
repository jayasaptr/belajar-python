```python
##classificaiton case study
df = pd.read_csv("telco_costumer.csv")
df
```

```python
df.info()
```

```python
##change data type
df = df[~(df["TotalCharges"] == " ")]

df["TotalCharges"] = df["TotalCharges"].astype("float64")
df.info()
```

```python
## one hot encoding
## adalah proses mengubah beberapa type object menjadi numerik tujuhnya agar bisa digunakan dalam model
df_obj = df.select_dtypes(["object"])
col_obj = list(df_obj.columns)
col_obj.remove("customerID")
col_obj.remove("Churn")
```

```python
df_clean = pd.get_dummies(df, columns=col_obj)
df_clean
```

```python
## train test split
## adalah proses membagi dataset menjadi dua bagian, yaitu data latih dan data uji
from sklearn.model_selection import train_test_split
```

```python
y = df_clean["Churn"].map({"Yes": 1, "No": 0})
x = df_clean.drop(columns=["Churn"])
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(x, y, test_size=0.3, random_state=1)
```

```python
print(X_train.shape)
print(X_test.shape)
print(Y_train.shape)
print(Y_test.shape)
```

```python
## imbalance data
## adalah kondisi di mana jumlah data pada setiap kelas tidak seimbang
Y_train.value_counts()
```

```python
from sklearn.utils import resample

df_train = pd.concat([X_train, Y_train], axis=1)

df_majority = df_train[df_train["Churn"] == 0]
df_minority = df_train[df_train["Churn"] == 1]

df_minority_upsampled = resample(df_minority, replace=True, n_samples=3608, random_state=1)

df_train_upsampled = pd.concat([df_majority, df_minority_upsampled])
```

```python
df_train_upsampled["Churn"].value_counts()
```

```python
Y_train_upsampled = df_train_upsampled["Churn"]
X_train_upsampled = df_train_upsampled.drop(columns=["Churn"])
```

```python
## remove column customerID
X_train = X_train.drop(columns=["customerID"])
X_train_upsampled = X_train_upsampled.drop(columns=["customerID"])
```

```python
## logistic regression
from sklearn.linear_model import LogisticRegression

log_reg = LogisticRegression()
log_reg.fit(X_train, Y_train)

log_reg_up = LogisticRegression()
log_reg_up.fit(X_train_upsampled, Y_train_upsampled)
```

```python
## decission tree
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier()
dt.fit(X_train, Y_train)

dt_up = DecisionTreeClassifier()
dt_up.fit(X_train_upsampled, Y_train_upsampled)
```

```python
## random forest
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier()
rf.fit(X_train, Y_train)

rf_up = RandomForestClassifier()
rf_up.fit(X_train_upsampled, Y_train_upsampled)
```

```python
params = {
    "max_depth": [2, 3, 5, 10, 15],
    "min_samples_leaf": [15, 30, 50, 100, 150]
}
```

```python
rf_base = RandomForestClassifier()
```

```python
from sklearn.model_selection import GridSearchCV

grid_search = GridSearchCV(estimator=rf_base, param_grid=params, cv=5, scoring="roc_auc")
```

```python
grid_search.fit(X_train, Y_train)
```

```python
rf_base = grid_search.best_estimator_
rf_base.fit(X_train, Y_train)
```

```python
grid_search.fit(X_train_upsampled, Y_train_upsampled)
```

```python
rf_base_up = grid_search.best_estimator_
rf_base_up.fit(X_train_upsampled, Y_train_upsampled)
```

```python
## prediction x-test
y_lr = log_reg.predict(X_test)
y_lr_up = log_reg_up.predict(X_test)
y_dt = dt.predict(X_test)
y_dt_up = dt_up.predict(X_test)
y_rf = rf.predict(X_test)
y_rf_up = rf_up.predict(X_test)
y_rf_base = rf_base.predict(X_test)
y_rf_base_up = rf_base_up.predict(X_test)
```

```python
## Evaluasi Model
```

```python
## precision
from sklearn.metrics import precision_score

precision_lr = precision_score(Y_test, y_lr, pos_label=1, average="binary")
precision_lr_up = precision_score(Y_test, y_lr_up, pos_label=1, average="binary")
precision_dt = precision_score(Y_test, y_dt, pos_label=1, average="binary")
precision_dt_up = precision_score(Y_test, y_dt_up, pos_label=1, average="binary")
precision_rf = precision_score(Y_test, y_rf, pos_label=1, average="binary")
precision_rf_up = precision_score(Y_test, y_rf_up, pos_label=1, average="binary")
precision_base = precision_score(Y_test, y_rf_base, pos_label=1, average="binary")
precision_base_up = precision_score(Y_test, y_rf_base_up, pos_label=1, average="binary")

print("precision lr:", precision_lr)
print("precision lr up:", precision_lr_up)
print("precision dt:", precision_dt)
print("precision dt up:", precision_dt_up)
print("precision rf:", precision_rf)
print("precision rf up:", precision_rf_up)
print("precision base:", precision_base)
print("precision base up:", precision_base_up)
```

```python
## recall
from sklearn.metrics import recall_score

recall_lr = recall_score(Y_test, y_lr, pos_label=1, average="binary")
recall_lr_up = recall_score(Y_test, y_lr_up, pos_label=1, average="binary")
recall_dt = recall_score(Y_test, y_dt, pos_label=1, average="binary")
recall_dt_up = recall_score(Y_test, y_dt_up, pos_label=1, average="binary")
recall_rf = recall_score(Y_test, y_rf, pos_label=1, average="binary")
recall_rf_up = recall_score(Y_test, y_rf_up, pos_label=1, average="binary")
recall_base = recall_score(Y_test, y_rf_base, pos_label=1, average="binary")
recall_base_up = recall_score(Y_test, y_rf_base_up, pos_label=1, average="binary")

print("recall lr:", recall_lr)
print("recall lr up:", recall_lr_up)
print("recall dt:", recall_dt)
print("recall dt up:", recall_dt_up)
print("recall rf:", recall_rf)
print("recall rf up:", recall_rf_up)
print("recall base:", recall_base)
print("recall base up:", recall_base_up)
```

```python
## AUC
y_lr = log_reg.predict_proba(X_test)[:, 1]
y_lr_up = log_reg_up.predict_proba(X_test)[:, 1]
y_dt = dt.predict_proba(X_test)[:, 1]
y_dt_up = dt_up.predict_proba(X_test)[:, 1]
y_rf = rf.predict_proba(X_test)[:, 1]
y_rf_up = rf_up.predict_proba(X_test)[:, 1]
y_rf_base = rf_base.predict_proba(X_test)[:, 1]
y_rf_base_up = rf_base_up.predict_proba(X_test)[:, 1]
```

```python
from sklearn.metrics import roc_auc_score
auc_lr = roc_auc_score(Y_test, y_lr)
auc_lr_up = roc_auc_score(Y_test, y_lr_up)
auc_dt = roc_auc_score(Y_test, y_dt)
auc_dt_up = roc_auc_score(Y_test, y_dt_up)
auc_rf = roc_auc_score(Y_test, y_rf)
auc_rf_up = roc_auc_score(Y_test, y_rf_up)
auc_base = roc_auc_score(Y_test, y_rf_base)
auc_base_up = roc_auc_score(Y_test, y_rf_base_up)

print("auc lr:", auc_lr)
print("auc lr up:", auc_lr_up)
print("auc dt:", auc_dt)
print("auc dt up:", auc_dt_up)
print("auc rf:", auc_rf)
print("auc rf up:", auc_rf_up)
print("auc base:", auc_base)
print("auc base up:", auc_base_up)
```
